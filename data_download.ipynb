```python
import os
import re
import time
import urllib.request
```

```python
# Dictionary of major Tolstoy works and their Project Gutenberg IDs
TOLSTOY_BOOKS = {
    "War_and_Peace": 2600,
    "Anna_Karenina": 1399,
    "The_Kingdom_of_God_is_Within_You": 43409,
    "Resurrection": 1938,
    "The_Death_of_Ivan_Ilyich": 28767,
    "The_Kreutzer_Sonata": 689,
    "Childhood": 19680,
    "Boyhood": 2450,
    "Youth": 2637,
    "Master_and_Man": 986
}
DOSTOEVSKY_BOOKS = {
    "The_Brothers_Karamazov": 28054,
    "Crime_and_Punishment": 2554,
    "The_Idiot": 2638,
    "Notes_from_the_Underground": 600,
    "The_Gambler": 2197,
    "The_House_of_the_Dead": 37536,
    "Poor_Folk": 2302,
    "White_Nights_and_Other_Stories": 36034,
    "The_Grand_Inquisitor": 8578,
}
```

```python
def clean_gutenberg_text(text):
    """Strips Project Gutenberg headers, footers, and formatting artifacts."""
    # Locating standard Project Gutenberg start and end markers
    start_match = re.search(
        r"\*\*\* START OF THE PROJECT GUTENBERG EBOOK .* \*\*\*", text
    )
    end_match = re.search(
        r"\*\*\* END OF THE PROJECT GUTENBERG EBOOK .* \*\*\*", text
    )

    if start_match:
        text = text[start_match.end() :]
    if end_match:
        text = text[: end_match.start()]

    # Remove carriage returns and trailing whitespace
    text = text.replace("\r\n", "\n")

    # Clean up excessive blank lines
    text = re.sub(r"\n{3,}", "\n\n", text)

    return text.strip()
```

```python
def download_tolstoy_corpus():
    OUTPUT_DIR = "data/tolstoy"
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    print(f"Starting download. Corpus will be saved to: {OUTPUT_DIR}/")

    for book_name, book_id in TOLSTOY_BOOKS.items():
        # Standard URL format for Project Gutenberg plain text files
        url = f"https://www.gutenberg.org/cache/epub/{book_id}/pg{book_id}.txt"
        print(f"Downloading {book_name} (ID: {book_id})...")

        try:
            # Request text with a standard User-Agent header to prevent being blocked
            req = urllib.request.Request(
                url, headers={"User-Agent": "Mozilla/5.0"}
            )
            with urllib.request.urlopen(req) as response:
                raw_text = response.read().decode("utf-8")

            # Clean headers/footers
            cleaned_text = clean_gutenberg_text(raw_text)

            # Save file
            file_path = os.path.join(OUTPUT_DIR, f"{book_name}.txt")
            with open(file_path, "w", encoding="utf-8") as f:
                f.write(cleaned_text)

            print(f"-> Successfully saved {book_name}.txt")

            # Polite scraping delay to avoid overwhelming Gutenberg servers
            time.sleep(2)

        except Exception as e:
            print(f"-> Failed to download {book_name}: {e}")

    print("\nDownload process complete!")
```
